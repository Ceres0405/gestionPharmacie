```python
# ==========================================
#        PERSEUS PHARMACIE
#        GESTION D'UNE OFFICINE
# ==========================================

import tkinter as tk
from tkinter import messagebox


# ==========================================
# BASE DE DONNÉES
# ==========================================

clients = []
produits = []
ventes = []


# ==========================================
# RECHERCHER
# ==========================================

def chercher_client(id_client):
    for client in clients:
        if client["id"] == id_client:
            return client
    return None


def chercher_produit(id_produit):
    for produit in produits:
        if produit["id"] == id_produit:
            return produit
    return None


# ==========================================
# CLIENTS
# ==========================================

def ajouter_client():
    id_client = entree_id_client.get()
    nom = entree_nom.get()
    prenom = entree_prenom.get()
    contact = entree_contact.get()

    if not id_client or not nom or not prenom or not contact:
        messagebox.showwarning(
            "PERSEUS PHARMACIE",
            "Veuillez remplir tous les champs."
        )
        return

    if chercher_client(id_client):
        messagebox.showerror(
            "Erreur",
            "Ce client existe déjà."
        )
        return

    clients.append({
        "id": id_client,
        "nom": nom,
        "prenom": prenom,
        "contact": contact
    })

    messagebox.showinfo(
        "PERSEUS PHARMACIE",
        "Client ajouté avec succès !"
    )

    entree_id_client.delete(0, tk.END)
    entree_nom.delete(0, tk.END)
    entree_prenom.delete(0, tk.END)
    entree_contact.delete(0, tk.END)


def afficher_clients():

    fenetre = tk.Toplevel(fenetre_principale)
    fenetre.title("PERSEUS PHARMACIE - Clients")
    fenetre.geometry("650x400")

    titre = tk.Label(
        fenetre,
        text="LISTE DES CLIENTS",
        font=("Arial", 18, "bold")
    )
    titre.pack(pady=15)

    zone = tk.Text(
        fenetre,
        width=75,
        height=18,
        font=("Arial", 11)
    )
    zone.pack(padx=20, pady=10)

    if not clients:
        zone.insert(tk.END, "Aucun client enregistré.")
    else:
        for client in clients:
            zone.insert(
                tk.END,
                f"ID : {client['id']} | "
                f"{client['nom']} {client['prenom']} | "
                f"Contact : {client['contact']}\n"
            )

    zone.config(state="disabled")


# ==========================================
# PRODUITS ET STOCK
# ==========================================

def ajouter_produit():

    id_produit = entree_id_produit.get()
    nom = entree_nom_produit.get()
    categorie = entree_categorie.get()

    try:
        prix = float(entree_prix.get())
        stock = int(entree_stock.get())
        seuil = int(entree_seuil.get())
    except ValueError:
        messagebox.showerror(
            "Erreur",
            "Prix, stock et seuil doivent être des nombres."
        )
        return

    if not id_produit or not nom or not categorie:
        messagebox.showwarning(
            "PERSEUS PHARMACIE",
            "Veuillez remplir tous les champs."
        )
        return

    if chercher_produit(id_produit):
        messagebox.showerror(
            "Erreur",
            "Ce produit existe déjà."
        )
        return

    if prix < 0 or stock < 0 or seuil < 0:
        messagebox.showerror(
            "Erreur",
            "Les valeurs ne peuvent pas être négatives."
        )
        return

    produits.append({
        "id": id_produit,
        "nom": nom,
        "categorie": categorie,
        "prix": prix,
        "stock": stock,
        "seuil": seuil
    })

    messagebox.showinfo(
        "PERSEUS PHARMACIE",
        "Produit ajouté avec succès !"
    )

    entree_id_produit.delete(0, tk.END)
    entree_nom_produit.delete(0, tk.END)
    entree_categorie.delete(0, tk.END)
    entree_prix.delete(0, tk.END)
    entree_stock.delete(0, tk.END)
    entree_seuil.delete(0, tk.END)


def afficher_produits():

    fenetre = tk.Toplevel(fenetre_principale)
    fenetre.title("PERSEUS PHARMACIE - Produits")
    fenetre.geometry("850x450")

    titre = tk.Label(
        fenetre,
        text="PRODUITS ET STOCK",
        font=("Arial", 18, "bold")
    )
    titre.pack(pady=15)

    zone = tk.Text(
        fenetre,
        width=100,
        height=20,
        font=("Arial", 10)
    )
    zone.pack(padx=20, pady=10)

    if not produits:
        zone.insert(tk.END, "Aucun produit enregistré.")
    else:
        for produit in produits:

            message = (
                f"ID : {produit['id']} | "
                f"Produit : {produit['nom']} | "
                f"Catégorie : {produit['categorie']} | "
                f"Prix : {produit['prix']} FCFA | "
                f"Stock : {produit['stock']} | "
                f"Seuil : {produit['seuil']}\n"
            )

            zone.insert(tk.END, message)

            if produit["stock"] <= produit["seuil"]:
                zone.insert(
                    tk.END,
                    "    ⚠ STOCK FAIBLE\n"
                )

    zone.config(state="disabled")


def ajouter_stock():

    id_produit = entree_id_stock.get()

    try:
        quantite = int(entree_quantite_stock.get())
    except ValueError:
        messagebox.showerror(
            "Erreur",
            "La quantité doit être un nombre entier."
        )
        return

    produit = chercher_produit(id_produit)

    if not produit:
        messagebox.showerror(
            "Erreur",
            "Produit introuvable."
        )
        return

    if quantite <= 0:
        messagebox.showerror(
            "Erreur",
            "La quantité doit être supérieure à zéro."
        )
        return

    produit["stock"] += quantite

    messagebox.showinfo(
        "PERSEUS PHARMACIE",
        f"Nouveau stock de {produit['nom']} : "
        f"{produit['stock']}"
    )

    entree_id_stock.delete(0, tk.END)
    entree_quantite_stock.delete(0, tk.END)


# ==========================================
# VENTES
# ==========================================

def enregistrer_vente():

    id_client = entree_id_vente_client.get()
    id_produit = entree_id_vente_produit.get()

    try:
        quantite = int(entree_quantite_vente.get())
    except ValueError:
        messagebox.showerror(
            "Erreur",
            "La quantité doit être un nombre entier."
        )
        return

    client = chercher_client(id_client)

    if not client:
        messagebox.showerror(
            "Erreur",
            "Client introuvable."
        )
        return

    produit = chercher_produit(id_produit)

    if not produit:
        messagebox.showerror(
            "Erreur",
            "Produit introuvable."
        )
        return

    if quantite <= 0:
        messagebox.showerror(
            "Erreur",
            "Quantité invalide."
        )
        return

    if quantite > produit["stock"]:
        messagebox.showerror(
            "Erreur",
            "Stock insuffisant."
        )
        return

    montant = quantite * produit["prix"]

    produit["stock"] -= quantite

    ventes.append({
        "client": id_client,
        "produit": id_produit,
        "quantite": quantite,
        "montant": montant
    })

    messagebox.showinfo(
        "PERSEUS PHARMACIE",
        f"Vente enregistrée !\n\n"
        f"Produit : {produit['nom']}\n"
        f"Quantité : {quantite}\n"
        f"Montant : {montant} FCFA\n"
        f"Nouveau stock : {produit['stock']}"
    )

    entree_id_vente_client.delete(0, tk.END)
    entree_id_vente_produit.delete(0, tk.END)
    entree_quantite_vente.delete(0, tk.END)


def afficher_ventes():

    fenetre = tk.Toplevel(fenetre_principale)
    fenetre.title("PERSEUS PHARMACIE - Ventes")
    fenetre.geometry("800x450")

    titre = tk.Label(
        fenetre,
        text="VENTES",
        font=("Arial", 18, "bold")
    )
    titre.pack(pady=15)

    zone = tk.Text(
        fenetre,
        width=90,
        height=20,
        font=("Arial", 10)
    )
    zone.pack(padx=20, pady=10)

    if not ventes:

        zone.insert(
            tk.END,
            "Aucune vente enregistrée."
        )

    else:

        total = 0

        for vente in ventes:

            client = chercher_client(
                vente["client"]
            )

            produit = chercher_produit(
                vente["produit"]
            )

            zone.insert(
                tk.END,
                f"Client : {client['nom']} {client['prenom']} | "
                f"Produit : {produit['nom']} | "
                f"Quantité : {vente['quantite']} | "
                f"Montant : {vente['montant']} FCFA\n"
            )

            total += vente["montant"]

        zone.insert(
            tk.END,
            "\n--------------------------------------\n"
        )

        zone.insert(
            tk.END,
            f"RECETTE TOTALE : {total} FCFA"
        )

    zone.config(state="disabled")


# ==========================================
# KPI
# ==========================================

def afficher_kpi():

    recette = 0
    quantite_totale = 0

    for vente in ventes:

        recette += vente["montant"]
        quantite_totale += vente["quantite"]

    nombre_ventes = len(ventes)

    if nombre_ventes > 0:
        moyenne = recette / nombre_ventes
    else:
        moyenne = 0

    fenetre = tk.Toplevel(fenetre_principale)
    fenetre.title("PERSEUS PHARMACIE - KPI")
    fenetre.geometry("600x400")

    titre = tk.Label(
        fenetre,
        text="INDICATEURS CLÉS DE PERFORMANCE",
        font=("Arial", 18, "bold")
    )
    titre.pack(pady=25)

    texte = (
        f"Nombre de ventes : {nombre_ventes}\n\n"
        f"Quantité totale vendue : {quantite_totale}\n\n"
        f"Recette totale : {recette} FCFA\n\n"
        f"Vente moyenne : {round(moyenne, 2)} FCFA"
    )

    label_kpi = tk.Label(
        fenetre,
        text=texte,
        font=("Arial", 14),
        justify="left"
    )

    label_kpi.pack(pady=20)


# ==========================================
# FENÊTRE CLIENTS
# ==========================================

def fenetre_clients():

    global entree_id_client
    global entree_nom
    global entree_prenom
    global entree_contact

    fenetre = tk.Toplevel(fenetre_principale)
    fenetre.title("PERSEUS PHARMACIE - Gestion des clients")
    fenetre.geometry("500x500")

    tk.Label(
        fenetre,
        text="GESTION DES CLIENTS",
        font=("Arial", 20, "bold")
    ).pack(pady=20)

    cadre = tk.Frame(fenetre)
    cadre.pack()

    tk.Label(cadre, text="ID client :").grid(
        row=0, column=0, pady=8
    )

    entree_id_client = tk.Entry(cadre)
    entree_id_client.grid(
        row=0, column=1, pady=8
    )

    tk.Label(cadre, text="Nom :").grid(
        row=1, column=0, pady=8
    )

    entree_nom = tk.Entry(cadre)
    entree_nom.grid(
        row=1, column=1, pady=8
    )

    tk.Label(cadre, text="Prénom :").grid(
        row=2, column=0, pady=8
    )

    entree_prenom = tk.Entry(cadre)
    entree_prenom.grid(
        row=2, column=1, pady=8
    )

    tk.Label(cadre, text="Contact :").grid(
        row=3, column=0, pady=8
    )

    entree_contact = tk.Entry(cadre)
    entree_contact.grid(
        row=3, column=1, pady=8
    )

    tk.Button(
        fenetre,
        text="Ajouter le client",
        command=ajouter_client,
        width=25
    ).pack(pady=15)

    tk.Button(
        fenetre,
        text="Afficher les clients",
        command=afficher_clients,
        width=25
    ).pack(pady=5)


# ==========================================
# FENÊTRE PRODUITS
# ==========================================

def fenetre_produits():

    global entree_id_produit
    global entree_nom_produit
    global entree_categorie
    global entree_prix
    global entree_stock
    global entree_seuil
    global entree_id_stock
    global entree_quantite_stock

    fenetre = tk.Toplevel(fenetre_principale)
    fenetre.title("PERSEUS PHARMACIE - Produits")
    fenetre.geometry("600x700")

    tk.Label(
        fenetre,
        text="GESTION DES PRODUITS ET DU STOCK",
        font=("Arial", 18, "bold")
    ).pack(pady=20)

    cadre = tk.Frame(fenetre)
    cadre.pack()

    tk.Label(cadre, text="ID produit :").grid(
        row=0, column=0, pady=7
    )

    entree_id_produit = tk.Entry(cadre)
    entree_id_produit.grid(
        row=0, column=1
    )

    tk.Label(cadre, text="Nom :").grid(
        row=1, column=0, pady=7
    )

    entree_nom_produit = tk.Entry(cadre)
    entree_nom_produit.grid(
        row=1, column=1
    )

    tk.Label(cadre, text="Catégorie :").grid(
        row=2, column=0, pady=7
    )

    entree_categorie = tk.Entry(cadre)
    entree_categorie.grid(
        row=2, column=1
    )

    tk.Label(cadre, text="Prix de vente :").grid(
        row=3, column=0, pady=7
    )

    entree_prix = tk.Entry(cadre)
    entree_prix.grid(
        row=3, column=1
    )

    tk.Label(cadre, text="Stock initial :").grid(
        row=4, column=0, pady=7
    )

    entree_stock = tk.Entry(cadre)
    entree_stock.grid(
        row=4, column=1
    )

    tk.Label(cadre, text="Seuil minimum :").grid(
        row=5, column=0, pady=7
    )

    entree_seuil = tk.Entry(cadre)
    entree_seuil.grid(
        row=5, column=1
    )

    tk.Button(
        fenetre,
        text="Ajouter le produit",
        command=ajouter_produit,
        width=30
    ).pack(pady=15)

    tk.Button(
        fenetre,
        text="Afficher les produits",
        command=afficher_produits,
        width=30
    ).pack(pady=5)

    tk.Label(
        fenetre,
        text="--- ENTRÉE DE STOCK ---",
        font=("Arial", 12, "bold")
    ).pack(pady=20)

    cadre_stock = tk.Frame(fenetre)
    cadre_stock.pack()

    tk.Label(
        cadre_stock,
        text="ID produit :"
    ).grid(row=0, column=0, pady=7)

    entree_id_stock = tk.Entry(cadre_stock)
    entree_id_stock.grid(row=0, column=1)

    tk.Label(
        cadre_stock,
        text="Quantité :"
    ).grid(row=1, column=0, pady=7)

    entree_quantite_stock = tk.Entry(cadre_stock)
    entree_quantite_stock.grid(row=1, column=1)

    tk.Button(
        fenetre,
        text="Ajouter au stock",
        command=ajouter_stock,
        width=30
    ).pack(pady=10)


# ==========================================
# FENÊTRE VENTES
# ==========================================

def fenetre_ventes():

    global entree_id_vente_client
    global entree_id_vente_produit
    global entree_quantite_vente

    fenetre = tk.Toplevel(fenetre_principale)
    fenetre.title("PERSEUS PHARMACIE - Ventes")
    fenetre.geometry("550x500")

    tk.Label(
        fenetre,
        text="ENREGISTRER UNE VENTE",
        font=("Arial", 20, "bold")
    ).pack(pady=25)

    cadre = tk.Frame(fenetre)
    cadre.pack()

    tk.Label(
        cadre,
        text="ID client :"
    ).grid(row=0, column=0, pady=10)

    entree_id_vente_client = tk.Entry(cadre)
    entree_id_vente_client.grid(
        row=0, column=1
    )

    tk.Label(
        cadre,
        text="ID produit :"
    ).grid(row=1, column=0, pady=10)

    entree_id_vente_produit = tk.Entry(cadre)
    entree_id_vente_produit.grid(
        row=1, column=1
    )

    tk.Label(
        cadre,
        text="Quantité :"
    ).grid(row=2, column=0, pady=10)

    entree_quantite_vente = tk.Entry(cadre)
    entree_quantite_vente.grid(
        row=2, column=1
    )

    tk.Button(
        fenetre,
        text="ENREGISTRER LA VENTE",
        command=enregistrer_vente,
        width=30,
        height=2
    ).pack(pady=30)

    tk.Button(
        fenetre,
        text="Afficher les ventes",
        command=afficher_ventes,
        width=30
    ).pack()


# ==========================================
# FENÊTRE PRINCIPALE
# ==========================================

def menu():

    global fenetre_principale

    fenetre_principale = tk.Tk()

    fenetre_principale.title(
        "PERSEUS PHARMACIE"
    )

    fenetre_principale.geometry(
        "700x650"
    )

    fenetre_principale.resizable(
        False,
        False
    )

    # --------------------------------------
    # TITRE
    # --------------------------------------

    titre = tk.Label(
        fenetre_principale,
        text="PERSEUS PHARMACIE",
        font=("Arial", 28, "bold")
    )

    titre.pack(pady=30)

    sous_titre = tk.Label(
        fenetre_principale,
        text="Système de gestion d'une officine pharmaceutique",
        font=("Arial", 13)
    )

    sous_titre.pack(pady=5)

    # --------------------------------------
    # BOUTONS
    # --------------------------------------

    cadre = tk.Frame(
        fenetre_principale
    )

    cadre.pack(pady=30)

    tk.Button(
        cadre,
        text="GESTION DES CLIENTS",
        command=fenetre_clients,
        width=35,
        height=2
    ).pack(pady=8)

    tk.Button(
        cadre,
        text="PRODUITS ET STOCK",
        command=fenetre_produits,
        width=35,
        height=2
    ).pack(pady=8)

    tk.Button(
        cadre,
        text="ENREGISTRER UNE VENTE",
        command=fenetre_ventes,
        width=35,
        height=2
    ).pack(pady=8)

    tk.Button(
        cadre,
        text="AFFICHER LES VENTES",
        command=afficher_ventes,
        width=35,
        height=2
    ).pack(pady=8)

    tk.Button(
        cadre,
        text="AFFICHER LES KPI",
        command=afficher_kpi,
        width=35,
        height=2
    ).pack(pady=8)

    tk.Button(
        cadre,
        text="QUITTER",
        command=fenetre_principale.destroy,
        width=35,
        height=2
    ).pack(pady=20)

    # --------------------------------------
    # PIED DE PAGE
    # --------------------------------------

    pied = tk.Label(
        fenetre_principale,
        text="PERSEUS PHARMACIE ©",
        font=("Arial", 10)
    )

    pied.pack(side="bottom", pady=15)

    fenetre_principale.mainloop()


# ==========================================
# LANCEMENT
# ==========================================

menu()
# ============================================================
#              PERSEUS PHARMACIE
#       GESTION D'UNE OFFICINE PHARMACEUTIQUE
# ============================================================


# ==========================================
# GESTION D'UNE OFFICINE PHARMACEUTIQUE
# ==========================================

clients = []
produits = []
ventes = []


# ==========================================
# RECHERCHER
# ==========================================

def chercher_client(id_client):
    for client in clients:
        if client["id"] == id_client:
            return client
    return None


def chercher_produit(id_produit):
    for produit in produits:
        if produit["id"] == id_produit:
            return produit
    return None


# ==========================================
# CLIENTS
# ==========================================

def ajouter_client():
    print("\n--- AJOUTER UN CLIENT ---")

    id_client = input("ID client : ")

    if chercher_client(id_client):
        print("Ce client existe déjà.")
        return

    nom = input("Nom : ")
    prenom = input("Prénom : ")
    contact = input("Contact : ")

    clients.append({
        "id": id_client,
        "nom": nom,
        "prenom": prenom,
        "contact": contact
    })

    print("Client ajouté !")


def afficher_clients():
    print("\n--- CLIENTS ---")

    if not clients:
        print("Aucun client.")
        return

    for client in clients:
        print(
            client["id"], "|",
            client["nom"], client["prenom"],
            "|", client["contact"]
        )


# ==========================================
# PRODUITS ET STOCK
# ==========================================

def ajouter_produit():
    print("\n--- AJOUTER UN PRODUIT ---")

    id_produit = input("ID produit : ")

    if chercher_produit(id_produit):
        print("Ce produit existe déjà.")
        return

    nom = input("Nom du produit : ")
    categorie = input("Catégorie : ")
    prix = float(input("Prix de vente : "))
    stock = int(input("Stock initial : "))
    seuil = int(input("Seuil minimum : "))

    produits.append({
        "id": id_produit,
        "nom": nom,
        "categorie": categorie,
        "prix": prix,
        "stock": stock,
        "seuil": seuil
    })

    print("Produit ajouté !")


def afficher_produits():
    print("\n--- PRODUITS ET STOCK ---")

    if not produits:
        print("Aucun produit.")
        return

    for produit in produits:
        print(
            produit["id"], "|",
            produit["nom"],
            "| Prix :", produit["prix"],
            "FCFA | Stock :", produit["stock"]
        )


def ajouter_stock():
    print("\n--- ENTRÉE DE STOCK ---")

    id_produit = input("ID produit : ")
    produit = chercher_produit(id_produit)

    if not produit:
        print("Produit introuvable.")
        return

    quantite = int(input("Quantité ajoutée : "))

    produit["stock"] += quantite

    print("Nouveau stock :", produit["stock"])


# ==========================================
# VENTES
# ==========================================

def enregistrer_vente():
    print("\n--- ENREGISTRER UNE VENTE ---")

    id_client = input("ID client : ")
    client = chercher_client(id_client)

    if not client:
        print("Client introuvable.")
        return

    id_produit = input("ID produit : ")
    produit = chercher_produit(id_produit)

    if not produit:
        print("Produit introuvable.")
        return

    print("Produit :", produit["nom"])
    print("Prix :", produit["prix"], "FCFA")
    print("Stock :", produit["stock"])

    quantite = int(input("Quantité vendue : "))

    if quantite <= 0:
        print("Quantité invalide.")
        return

    if quantite > produit["stock"]:
        print("Stock insuffisant.")
        return

    montant = quantite * produit["prix"]

    produit["stock"] -= quantite

    ventes.append({
        "client": id_client,
        "produit": id_produit,
        "quantite": quantite,
        "montant": montant
    })

    print("\nVente enregistrée !")
    print("Montant :", montant, "FCFA")
    print("Nouveau stock :", produit["stock"])


def afficher_ventes():
    print("\n--- VENTES ---")

    if not ventes:
        print("Aucune vente.")
        return

    total = 0

    for vente in ventes:

        produit = chercher_produit(vente["produit"])

        print(
            "Produit :", produit["nom"],
            "| Quantité :", vente["quantite"],
            "| Montant :", vente["montant"],
            "FCFA"
        )

        total += vente["montant"]

    print("-------------------------")
    print("RECETTE TOTALE :", total, "FCFA")


# ==========================================
# KPI
# ==========================================

def afficher_kpi():
    print("\n--- KPI DE L'OFFICINE ---")

    recette = 0
    quantite_totale = 0

    for vente in ventes:
        recette += vente["montant"]
        quantite_totale += vente["quantite"]

    print("Nombre de ventes :", len(ventes))
    print("Quantité vendue :", quantite_totale)
    print("Recette totale :", recette, "FCFA")

    if ventes:
        moyenne = recette / len(ventes)
        print("Vente moyenne :", round(moyenne, 2), "FCFA")
    else:
        print("Vente moyenne : 0 FCFA")


# ==========================================
# MENU PRINCIPAL
# ==========================================

def menu():

    while True:

        print("\n================================")
        print("       PERSEUS PHARMACIE")
        print("================================")

        print("1 - Gestion des clients")
        print("2 - Gestion des produits")
        print("3 - Enregistrer une vente")
        print("4 - Afficher les ventes")
        print("5 - Afficher les KPI")
        print("0 - Quitter")

        choix = input("\nVotre choix : ")

        if choix == "1":

            print("\n1 - Ajouter un client")
            print("2 - Afficher les clients")

            sous_choix = input("Votre choix : ")

            if sous_choix == "1":
                ajouter_client()

            elif sous_choix == "2":
                afficher_clients()

            else:
                print("Choix invalide.")

        elif choix == "2":

            print("\n1 - Ajouter un produit")
            print("2 - Afficher les produits")
            print("3 - Ajouter du stock")

            sous_choix = input("Votre choix : ")

            if sous_choix == "1":
                ajouter_produit()

            elif sous_choix == "2":
                afficher_produits()

            elif sous_choix == "3":
                ajouter_stock()

            else:
                print("Choix invalide.")

        elif choix == "3":
            enregistrer_vente()

        elif choix == "4":
            afficher_ventes()

        elif choix == "5":
            afficher_kpi()

        elif choix == "0":
            print("\nProgramme terminé.")
            break

        else:
            print("Choix invalide.")


# ============================================================
#              INTERFACE GRAPHIQUE
# ============================================================

import tkinter as tk
from tkinter import ttk, messagebox
import builtins
from contextlib import redirect_stdout
from io import StringIO


# ============================================================
# COULEURS
# ============================================================

FOND = "#F4F7F6"
VERT_FONCE = "#123C35"
VERT = "#21A67A"
VERT_CLAIR = "#E3F5EE"
BLANC = "#FFFFFF"
TEXTE = "#18332E"
GRIS = "#71827E"
ROUGE = "#D9534F"
VIOLET = "#8067D8"
BLEU = "#4285B8"


# ============================================================
# OUTIL POUR UTILISER LES FONCTIONS ORIGINALES
# ============================================================

def executer_fonction(fonction, reponses):

    ancien_input = builtins.input

    position = 0

    def faux_input(message=""):
        nonlocal position

        valeur = reponses[position]
        position += 1

        return str(valeur)

    builtins.input = faux_input

    sortie = StringIO()

    try:

        with redirect_stdout(sortie):
            fonction()

    finally:

        builtins.input = ancien_input

    return sortie.getvalue()


# ============================================================
# APPLICATION
# ============================================================

class PerseusPharmacie(tk.Tk):

    def __init__(self):

        super().__init__()

        self.title("PERSEUS PHARMACIE")

        self.geometry("1200x720")

        self.minsize(1000, 650)

        self.configure(bg=FOND)

        style = ttk.Style()

        style.theme_use("clam")

        style.configure(
            "Treeview",
            background=BLANC,
            foreground=TEXTE,
            fieldbackground=BLANC,
            rowheight=36,
            font=("Arial", 10)
        )

        style.configure(
            "Treeview.Heading",
            background=VERT_FONCE,
            foreground=BLANC,
            font=("Arial", 10, "bold")
        )

        self.creer_interface()

        self.dashboard()


    # ========================================================
    # INTERFACE PRINCIPALE
    # ========================================================

    def creer_interface(self):

        self.menu_gauche = tk.Frame(
            self,
            bg=VERT_FONCE,
            width=240
        )

        self.menu_gauche.pack(
            side="left",
            fill="y"
        )

        self.menu_gauche.pack_propagate(False)


        self.zone = tk.Frame(
            self,
            bg=FOND
        )

        self.zone.pack(
            side="right",
            fill="both",
            expand=True
        )


        # LOGO

        tk.Label(
            self.menu_gauche,
            text="✚",
            font=("Arial", 35, "bold"),
            fg=VERT,
            bg=VERT_FONCE
        ).pack(
            pady=(30, 0)
        )


        tk.Label(
            self.menu_gauche,
            text="PERSEUS",
            font=("Arial", 21, "bold"),
            fg=BLANC,
            bg=VERT_FONCE
        ).pack()


        tk.Label(
            self.menu_gauche,
            text="PHARMACIE",
            font=("Arial", 12, "bold"),
            fg=VERT,
            bg=VERT_FONCE
        ).pack(
            pady=(0, 30)
        )


        # BOUTONS

        self.bouton_menu(
            "⌂   Tableau de bord",
            self.dashboard
        )

        self.bouton_menu(
            "♙   Clients",
            self.page_clients
        )

        self.bouton_menu(
            "▣   Produits & stock",
            self.page_produits
        )

        self.bouton_menu(
            "◆   Nouvelle vente",
            self.page_vente
        )

        self.bouton_menu(
            "▤   Ventes",
            self.page_ventes
        )

        self.bouton_menu(
            "◉   KPI",
            self.page_kpi
        )


        tk.Label(
            self.menu_gauche,
            text="Gestion d'officine",
            font=("Arial", 9),
            fg="#78958E",
            bg=VERT_FONCE
        ).pack(
            side="bottom",
            pady=25
        )


    def bouton_menu(self, texte, commande):

        bouton = tk.Button(
            self.menu_gauche,
            text=texte,
            command=commande,
            anchor="w",
            bg=VERT_FONCE,
            fg=BLANC,
            activebackground="#1E554B",
            activeforeground=BLANC,
            bd=0,
            font=("Arial", 10, "bold"),
            padx=20,
            pady=14,
            cursor="hand2"
        )

        bouton.pack(
            fill="x",
            padx=10,
            pady=2
        )


    # ========================================================
    # NETTOYER LA PAGE
    # ========================================================

    def nettoyer(self):

        for widget in self.zone.winfo_children():
            widget.destroy()


    # ========================================================
    # TITRE
    # ========================================================

    def titre(self, titre, sous_titre=""):

        bloc = tk.Frame(
            self.zone,
            bg=FOND
        )

        bloc.pack(
            fill="x",
            padx=30,
            pady=(25, 20)
        )

        tk.Label(
            bloc,
            text=titre,
            font=("Arial", 24, "bold"),
            fg=TEXTE,
            bg=FOND
        ).pack(
            anchor="w"
        )

        tk.Label(
            bloc,
            text=sous_titre,
            font=("Arial", 10),
            fg=GRIS,
            bg=FOND
        ).pack(
            anchor="w",
            pady=4
        )


    # ========================================================
    # DASHBOARD
    # ========================================================

    def dashboard(self):

        self.nettoyer()

        self.titre(
            "Bonjour 👋",
            "Bienvenue dans votre espace de gestion — PERSEUS PHARMACIE"
        )


        recette = sum(
            vente["montant"]
            for vente in ventes
        )

        nombre_ventes = len(ventes)

        quantite = sum(
            vente["quantite"]
            for vente in ventes
        )

        alertes = sum(
            1
            for produit in produits
            if produit["stock"] <= produit["seuil"]
        )


        cartes = tk.Frame(
            self.zone,
            bg=FOND
        )

        cartes.pack(
            fill="x",
            padx=30
        )


        self.creer_carte(
            cartes,
            "RECETTE TOTALE",
            f"{recette:,.0f} FCFA",
            VERT
        ).pack(
            side="left",
            fill="both",
            expand=True,
            padx=5
        )


        self.creer_carte(
            cartes,
            "NOMBRE DE VENTES",
            str(nombre_ventes),
            VIOLET
        ).pack(
            side="left",
            fill="both",
            expand=True,
            padx=5
        )


        self.creer_carte(
            cartes,
            "QUANTITÉ VENDUE",
            str(quantite),
            BLEU
        ).pack(
            side="left",
            fill="both",
            expand=True,
            padx=5
        )


        self.creer_carte(
            cartes,
            "STOCKS EN ALERTE",
            str(alertes),
            ROUGE if alertes else VERT
        ).pack(
            side="left",
            fill="both",
            expand=True,
            padx=5
        )


        # ACTIONS RAPIDES

        cadre = tk.Frame(
            self.zone,
            bg=BLANC,
            highlightbackground="#DCE6E2",
            highlightthickness=1
        )

        cadre.pack(
            fill="x",
            padx=30,
            pady=25
        )


        tk.Label(
            cadre,
            text="⚡ Actions rapides",
            font=("Arial", 15, "bold"),
            fg=TEXTE,
            bg=BLANC
        ).pack(
            anchor="w",
            padx=20,
            pady=15
        )


        boutons = tk.Frame(
            cadre,
            bg=BLANC
        )

        boutons.pack(
            fill="x",
            padx=20,
            pady=(0, 20)
        )


        tk.Button(
            boutons,
            text="＋ Ajouter un client",
            command=self.ajouter_client_gui,
            bg=VERT,
            fg=BLANC,
            bd=0,
            cursor="hand2",
            font=("Arial", 10, "bold"),
            padx=18,
            pady=12
        ).pack(
            side="left",
            padx=5
        )


        tk.Button(
            boutons,
            text="＋ Ajouter un produit",
            command=self.ajouter_produit_gui,
            bg=BLEU,
            fg=BLANC,
            bd=0,
            cursor="hand2",
            font=("Arial", 10, "bold"),
            padx=18,
            pady=12
        ).pack(
            side="left",
            padx=5
        )


        tk.Button(
            boutons,
            text="◆ Nouvelle vente",
            command=self.page_vente,
            bg=VIOLET,
            fg=BLANC,
            bd=0,
            cursor="hand2",
            font=("Arial", 10, "bold"),
            padx=18,
            pady=12
        ).pack(
            side="left",
            padx=5
        )


    # ========================================================
    # CARTE
    # ========================================================

    def creer_carte(
        self,
        parent,
        titre,
        valeur,
        couleur
    ):

        cadre = tk.Frame(
            parent,
            bg=BLANC,
            highlightbackground="#DCE6E2",
            highlightthickness=1
        )

        tk.Label(
            cadre,
            text=titre,
            font=("Arial", 9, "bold"),
            fg=GRIS,
            bg=BLANC
        ).pack(
            anchor="w",
            padx=18,
            pady=(18, 5)
        )

        tk.Label(
            cadre,
            text=valeur,
            font=("Arial", 17, "bold"),
            fg=couleur,
            bg=BLANC
        ).pack(
            anchor="w",
            padx=18,
            pady=(0, 18)
        )

        return cadre


    # ========================================================
    # CLIENTS
    # ========================================================

    def page_clients(self):

        self.nettoyer()

        self.titre(
            "Clients",
            "Gestion des clients de PERSEUS PHARMACIE"
        )


        tk.Button(
            self.zone,
            text="＋ Ajouter un client",
            command=self.ajouter_client_gui,
            bg=VERT,
            fg=BLANC,
            bd=0,
            cursor="hand2",
            font=("Arial", 10, "bold"),
            padx=18,
            pady=10
        ).pack(
            anchor="w",
            padx=30,
            pady=(0, 15)
        )


        cadre = tk.Frame(
            self.zone,
            bg=BLANC
        )

        cadre.pack(
            fill="both",
            expand=True,
            padx=30,
            pady=(0, 25)
        )


        tableau = ttk.Treeview(
            cadre,
            columns=(
                "id",
                "nom",
                "prenom",
                "contact"
            ),
            show="headings"
        )


        for colonne, texte in [
            ("id", "ID"),
            ("nom", "Nom"),
            ("prenom", "Prénom"),
            ("contact", "Contact")
        ]:

            tableau.heading(
                colonne,
                text=texte
            )


        tableau.pack(
            fill="both",
            expand=True,
            padx=10,
            pady=10
        )


        for client in clients:

            tableau.insert(
                "",
                "end",
                values=(
                    client["id"],
                    client["nom"],
                    client["prenom"],
                    client["contact"]
                )
            )


    def ajouter_client_gui(self):

        fenetre = tk.Toplevel(self)

        fenetre.title(
            "Ajouter un client"
        )

        fenetre.geometry(
            "420x420"
        )

        fenetre.configure(
            bg=BLANC
        )

        fenetre.grab_set()


        tk.Label(
            fenetre,
            text="NOUVEAU CLIENT",
            font=("Arial", 18, "bold"),
            fg=TEXTE,
            bg=BLANC
        ).pack(
            pady=20
        )


        champs = []


        for texte in [
            "ID client",
            "Nom",
            "Prénom",
            "Contact"
        ]:

            tk.Label(
                fenetre,
                text=texte,
                font=("Arial", 9, "bold"),
                fg=TEXTE,
                bg=BLANC
            ).pack(
                anchor="w",
                padx=30
            )


            entree = tk.Entry(
                fenetre,
                font=("Arial", 10)
            )

            entree.pack(
                fill="x",
                padx=30,
                pady=(3, 10),
                ipady=6
            )

            champs.append(entree)


        def enregistrer():

            reponses = [
                champ.get()
                for champ in champs
            ]


            sortie = executer_fonction(
                ajouter_client,
                reponses
            )


            if "existe déjà" in sortie:

                messagebox.showerror(
                    "Erreur",
                    "Ce client existe déjà.",
                    parent=fenetre
                )

                return


            fenetre.destroy()

            messagebox.showinfo(
                "Succès",
                "Client ajouté avec succès !"
            )

            self.page_clients()


        tk.Button(
            fenetre,
            text="ENREGISTRER",
            command=enregistrer,
            bg=VERT,
            fg=BLANC,
            bd=0,
            cursor="hand2",
            font=("Arial", 10, "bold"),
            padx=20,
            pady=10
        ).pack(
            pady=10
        )


    # ========================================================
    # PRODUITS
    # ========================================================

    def page_produits(self):

        self.nettoyer()

        self.titre(
            "Produits & Stock",
            "Gestion des médicaments et du stock"
        )


        boutons = tk.Frame(
            self.zone,
            bg=FOND
        )

        boutons.pack(
            fill="x",
            padx=30,
            pady=(0, 15)
        )


        tk.Button(
            boutons,
            text="＋ Ajouter un produit",
            command=self.ajouter_produit_gui,
            bg=VERT,
            fg=BLANC,
            bd=0,
            cursor="hand2",
            font=("Arial", 10, "bold"),
            padx=18,
            pady=10
        ).pack(
            side="left"
        )


        tk.Button(
            boutons,
            text="＋ Ajouter du stock",
            command=self.ajouter_stock_gui,
            bg=VIOLET,
            fg=BLANC,
            bd=0,
            cursor="hand2",
            font=("Arial", 10, "bold"),
            padx=18,
            pady=10
        ).pack(
            side="left",
            padx=10
        )


        cadre = tk.Frame(
            self.zone,
            bg=BLANC
        )

        cadre.pack(
            fill="both",
            expand=True,
            padx=30,
            pady=(0, 25)
        )


        tableau = ttk.Treeview(
            cadre,
            columns=(
                "id",
                "nom",
                "categorie",
                "prix",
                "stock",
                "seuil"
            ),
            show="headings"
        )


        for colonne, texte in [
            ("id", "ID"),
            ("nom", "Produit"),
            ("categorie", "Catégorie"),
            ("prix", "Prix"),
            ("stock", "Stock"),
            ("seuil", "Seuil")
        ]:

            tableau.heading(
                colonne,
                text=texte
            )


        tableau.pack(
            fill="both",
            expand=True,
            padx=10,
            pady=10
        )


        for produit in produits:

            tableau.insert(
                "",
                "end",
                values=(
                    produit["id"],
                    produit["nom"],
                    produit["categorie"],
                    f"{produit['prix']:,.0f} FCFA",
                    produit["stock"],
                    produit["seuil"]
                )
            )


    def ajouter_produit_gui(self):

        fenetre = tk.Toplevel(self)

        fenetre.title(
            "Ajouter un produit"
        )

        fenetre.geometry(
            "450x560"
        )

        fenetre.configure(
            bg=BLANC
        )

        fenetre.grab_set()


        tk.Label(
            fenetre,
            text="NOUVEAU PRODUIT",
            font=("Arial", 18, "bold"),
            fg=TEXTE,
            bg=BLANC
        ).pack(
            pady=20
        )


        champs = []


        for texte in [
            "ID produit",
            "Nom du produit",
            "Catégorie",
            "Prix de vente",
            "Stock initial",
            "Seuil minimum"
        ]:

            tk.Label(
                fenetre,
                text=texte,
                font=("Arial", 9, "bold"),
                fg=TEXTE,
                bg=BLANC
            ).pack(
                anchor="w",
                padx=30
            )


            entree = tk.Entry(
                fenetre,
                font=("Arial", 10)
            )

            entree.pack(
                fill="x",
                padx=30,
                pady=(3, 8),
                ipady=6
            )

            champs.append(entree)


        def enregistrer():

            try:

                float(champs[3].get())
                int(champs[4].get())
                int(champs[5].get())

            except ValueError:

                messagebox.showerror(
                    "Erreur",
                    "Prix, stock et seuil doivent être numériques.",
                    parent=fenetre
                )

                return


            sortie = executer_fonction(
                ajouter_produit,
                [
                    champ.get()
                    for champ in champs
                ]
            )


            if "existe déjà" in sortie:

                messagebox.showerror(
                    "Erreur",
                    "Ce produit existe déjà.",
                    parent=fenetre
                )

                return


            fenetre.destroy()

            messagebox.showinfo(
                "Succès",
                "Produit ajouté avec succès !"
            )

            self.page_produits()


        tk.Button(
            fenetre,
            text="ENREGISTRER",
            command=enregistrer,
            bg=VERT,
            fg=BLANC,
            bd=0,
            cursor="hand2",
            font=("Arial", 10, "bold"),
            padx=20,
            pady=10
        ).pack(
            pady=15
        )


    def ajouter_stock_gui(self):

        if not produits:

            messagebox.showinfo(
                "Stock",
                "Aucun produit n'est enregistré."
            )

            return


        fenetre = tk.Toplevel(self)

        fenetre.title(
            "Ajouter du stock"
        )

        fenetre.geometry(
            "420x300"
        )

        fenetre.configure(
            bg=BLANC
        )

        fenetre.grab_set()


        tk.Label(
            fenetre,
            text="ENTRÉE DE STOCK",
            font=("Arial", 18, "bold"),
            fg=TEXTE,
            bg=BLANC
        ).pack(
            pady=20
        )


        tk.Label(
            fenetre,
            text="Produit",
            font=("Arial", 9, "bold"),
            fg=TEXTE,
            bg=BLANC
        ).pack(
            anchor="w",
            padx=30
        )


        choix = ttk.Combobox(
            fenetre,
            state="readonly",
            values=[
                f"{p['id']} - {p['nom']}"
                for p in produits
            ]
        )

        choix.pack(
            fill="x",
            padx=30,
            ipady=6
        )

        choix.current(0)


        tk.Label(
            fenetre,
            text="Quantité ajoutée",
            font=("Arial", 9, "bold"),
            fg=TEXTE,
            bg=BLANC
        ).pack(
            anchor="w",
            padx=30,
            pady=(15, 3)
        )


        quantite = tk.Entry(
            fenetre
        )

        quantite.pack(
            fill="x",
            padx=30,
            ipady=6
        )


        def ajouter():

            try:
                int(quantite.get())

            except ValueError:

                messagebox.showerror(
                    "Erreur",
                    "La quantité doit être un nombre entier.",
                    parent=fenetre
                )

                return


            produit = produits[
                choix.current()
            ]


            sortie = executer_fonction(
                ajouter_stock,
                [
                    produit["id"],
                    quantite.get()
                ]
            )


            fenetre.destroy()

            messagebox.showinfo(
                "Stock",
                sortie.strip()
            )

            self.page_produits()


        tk.Button(
            fenetre,
            text="AJOUTER",
            command=ajouter,
            bg=VIOLET,
            fg=BLANC,
            bd=0,
            cursor="hand2",
            font=("Arial", 10, "bold"),
            padx=25,
            pady=10
        ).pack(
            pady=18
        )


    # ========================================================
    # VENTE
    # ========================================================

    def page_vente(self):

        self.nettoyer()

        self.titre(
            "Nouvelle vente",
            "Enregistrer une vente dans PERSEUS PHARMACIE"
        )


        if not clients:

            tk.Label(
                self.zone,
                text="Aucun client enregistré.",
                font=("Arial", 14),
                fg=ROUGE,
                bg=FOND
            ).pack(
                pady=50
            )

            return


        if not produits:

            tk.Label(
                self.zone,
                text="Aucun produit enregistré.",
                font=("Arial", 14),
                fg=ROUGE,
                bg=FOND
            ).pack(
                pady=50
            )

            return


        cadre = tk.Frame(
            self.zone,
            bg=BLANC,
            highlightbackground="#DCE6E2",
            highlightthickness=1
        )

        cadre.pack(
            fill="x",
            padx=30,
            pady=10
        )


        tk.Label(
            cadre,
            text="Client",
            font=("Arial", 10, "bold"),
            fg=TEXTE,
            bg=BLANC
        ).pack(
            anchor="w",
            padx=30,
            pady=(25, 5)
        )


        client_box = ttk.Combobox(
            cadre,
            state="readonly",
            values=[
                f"{c['id']} - {c['nom']} {c['prenom']}"
                for c in clients
            ]
        )

        client_box.pack(
            fill="x",
            padx=30,
            ipady=7
        )

        client_box.current(0)


        tk.Label(
            cadre,
            text="Produit",
            font=("Arial", 10, "bold"),
            fg=TEXTE,
            bg=BLANC
        ).pack(
            anchor="w",
            padx=30,
            pady=(20, 5)
        )


        produit_box = ttk.Combobox(
            cadre,
            state="readonly",
            values=[
                f"{p['id']} - {p['nom']}"
                for p in produits
            ]
        )

        produit_box.pack(
            fill="x",
            padx=30,
            ipady=7
        )

        produit_box.current(0)


        tk.Label(
            cadre,
            text="Quantité",
            font=("Arial", 10, "bold"),
            fg=TEXTE,
            bg=BLANC
        ).pack(
            anchor="w",
            padx=30,
            pady=(20, 5)
        )


        quantite = tk.Entry(
            cadre,
            font=("Arial", 11)
        )

        quantite.pack(
            anchor="w",
            padx=30,
            ipady=7
        )

        quantite.insert(
            0,
            "1"
        )


        information = tk.Label(
            cadre,
            text="",
            font=("Arial", 11),
            fg=TEXTE,
            bg=VERT_CLAIR,
            padx=15,
            pady=15
        )

        information.pack(
            fill="x",
            padx=30,
            pady=20
        )


        def actualiser():

            produit = produits[
                produit_box.current()
            ]

            information.config(
                text=(
                    f"Produit : {produit['nom']}    |    "
                    f"Prix : {produit['prix']:,.0f} FCFA    |    "
                    f"Stock disponible : {produit['stock']}"
                )
            )


        produit_box.bind(
            "<<ComboboxSelected>>",
            lambda event: actualiser()
        )


        actualiser()


        def vendre():

            client = clients[
                client_box.current()
            ]

            produit = produits[
                produit_box.current()
            ]


            sortie = executer_fonction(
                enregistrer_vente,
                [
                    client["id"],
                    produit["id"],
                    quantite.get()
                ]
            )


            if "introuvable" in sortie.lower():

                messagebox.showerror(
                    "Erreur",
                    sortie.strip()
                )

                return


            if "insuffisant" in sortie.lower():

                messagebox.showerror(
                    "Stock insuffisant",
                    sortie.strip()
                )

                return


            if "invalide" in sortie.lower():

                messagebox.showerror(
                    "Erreur",
                    sortie.strip()
                )

                return


            messagebox.showinfo(
                "Vente enregistrée",
                sortie.strip()
            )

            self.page_vente()


        tk.Button(
            cadre,
            text="◆  ENREGISTRER LA VENTE",
            command=vendre,
            bg=VERT,
            fg=BLANC,
            activebackground="#178665",
            bd=0,
            cursor="hand2",
            font=("Arial", 11, "bold"),
            padx=25,
            pady=12
        ).pack(
            pady=(5, 30)
        )


    # ========================================================
    # VENTES
    # ========================================================

    def page_ventes(self):

        self.nettoyer()

        self.titre(
            "Historique des ventes",
            "Toutes les ventes enregistrées"
        )


        recette = sum(
            vente["montant"]
            for vente in ventes
        )


        tk.Label(
            self.zone,
            text=f"Recette totale : {recette:,.0f} FCFA",
            font=("Arial", 16, "bold"),
            fg=VERT,
            bg=FOND
        ).pack(
            anchor="w",
            padx=30,
            pady=(0, 15)
        )


        cadre = tk.Frame(
            self.zone,
            bg=BLANC
        )

        cadre.pack(
            fill="both",
            expand=True,
            padx=30,
            pady=(0, 25)
        )


        tableau = ttk.Treeview(
            cadre,
            columns=(
                "client",
                "produit",
                "quantite",
                "montant"
            ),
            show="headings"
        )


        for colonne, texte in [
            ("client", "Client"),
            ("produit", "Produit"),
            ("quantite", "Quantité"),
            ("montant", "Montant")
        ]:

            tableau.heading(
                colonne,
                text=texte
            )


        tableau.pack(
            fill="both",
            expand=True,
            padx=10,
            pady=10
        )


        for vente in ventes:

            client = chercher_client(
                vente["client"]
            )

            produit = chercher_produit(
                vente["produit"]
            )


            nom_client = (
                f"{client['nom']} {client['prenom']}"
                if client
                else vente["client"]
            )


            nom_produit = (
                produit["nom"]
                if produit
                else vente["produit"]
            )


            tableau.insert(
                "",
                "end",
                values=(
                    nom_client,
                    nom_produit,
                    vente["quantite"],
                    f"{vente['montant']:,.0f} FCFA"
                )
            )


    # ========================================================
    # KPI
    # ========================================================

    def page_kpi(self):

        self.nettoyer()

        self.titre(
            "KPI de PERSEUS PHARMACIE",
            "Indicateurs de performance de l'officine"
        )


        recette = 0
        quantite = 0


        for vente in ventes:

            recette += vente["montant"]

            quantite += vente["quantite"]


        if ventes:
            moyenne = recette / len(ventes)

        else:
            moyenne = 0


        cartes = tk.Frame(
            self.zone,
            bg=FOND
        )

        cartes.pack(
            fill="x",
            padx=30
        )


        self.creer_carte(
            cartes,
            "NOMBRE DE VENTES",
            str(len(ventes)),
            VIOLET
        ).pack(
            side="left",
            fill="both",
            expand=True,
            padx=5
        )


        self.creer_carte(
            cartes,
            "QUANTITÉ VENDUE",
            str(quantite),
            BLEU
        ).pack(
            side="left",
            fill="both",
            expand=True,
            padx=5
        )


        self.creer_carte(
            cartes,
            "RECETTE TOTALE",
            f"{recette:,.0f} FCFA",
            VERT
        ).pack(
            side="left",
            fill="both",
            expand=True,
            padx=5
        )


        self.creer_carte(
            cartes,
            "VENTE MOYENNE",
            f"{moyenne:,.2f} FCFA",
            ROUGE
        ).pack(
            side="left",
            fill="both",
            expand=True,
            padx=5
        )


        # Résumé

        cadre = tk.Frame(
            self.zone,
            bg=BLANC,
            highlightbackground="#DCE6E2",
            highlightthickness=1
        )

        cadre.pack(
            fill="x",
            padx=30,
            pady=30
        )


        tk.Label(
            cadre,
            text="Résumé des performances",
            font=("Arial", 16, "bold"),
            fg=TEXTE,
            bg=BLANC
        ).pack(
            anchor="w",
            padx=25,
            pady=20
        )


        informations = [
            ("Nombre de ventes", len(ventes)),
            ("Quantité totale vendue", quantite),
            ("Recette totale", f"{recette:,.2f} FCFA"),
            ("Vente moyenne", f"{moyenne:,.2f} FCFA")
        ]


        for nom, valeur in informations:

            ligne = tk.Frame(
                cadre,
                bg=BLANC
            )

            ligne.pack(
                fill="x",
                padx=25
            )


            tk.Label(
                ligne,
                text=nom,
                font=("Arial", 10),
                fg=TEXTE,
                bg=BLANC
            ).pack(
                side="left",
                pady=10
            )


            tk.Label(
                ligne,
                text=str(valeur),
                font=("Arial", 10, "bold"),
                fg=VERT,
                bg=BLANC
            ).pack(
                side="right",
                pady=10
            )


# ============================================================
# LANCEMENT
# ============================================================

application = PerseusPharmacie()

application.mainloop()
 
 
```
